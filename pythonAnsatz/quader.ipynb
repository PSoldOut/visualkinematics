```python

import pythreejs as three
from pythreejs import *
from ipywidgets import *
from IPython.display import display, Javascript
import numpy as np
import util



def toggle_grid(change):
    grid_group.visible = not grid_group.visible

def toggle_axes(change):
    axes_group.visible = not axes_group.visible


  
width = 700
height = 500

scene = Scene()
scene.background = "#DDDDDD"


# Licht und Kamera
light = PointLight(color='white', intensity=1.5, position=[5, 5, 5])
camera = PerspectiveCamera(position=[8, 8, 8],aspect=width/height, fov=50)
camera.up = [0, 0, 1]

grid_group = util.create_grid_XY(14,0.5)

axes_group = util.create_axes(8, name="B")

b = three.Group()
k = util.create_axes(3, show_labels=True, name="K")

rectangle = util.createQuad([0,0,0],4,2,3)
k.add(rectangle)
b.add(k)

mat = np.array([
    [ 0,  0, -1],
    [-1,  0,  0],
    [ 0,  1,  0]
])
euler = util.rot_matrix_to_euler(mat, "XYZ")
scene.add([camera, light, axes_group, grid_group, b, AmbientLight(intensity=0.5)])

util.translate(k, [3, 4, 0])
util.translate(rectangle, [2, 1, 1.5])



# Renderer mit Orbit-Steuerung
renderer = Renderer(camera=camera, scene=scene, controls=[OrbitControls(controlling=camera)], width=width, height=height, background_color="#87CEEB", background_opacity=1.0, antialias=True, precision='highp')


checkbox_grid = Checkbox(value=True, description='Show Grid')
checkbox_axes = Checkbox(value=True, description='Show Axes')


#interactive_control_scale = widgets.interactive(update_cube_scale, x=x_scale_slider, y=y_scale_slider, z=z_scale_slider)
checkbox_grid.observe(toggle_grid, names='value')
checkbox_axes.observe(toggle_axes, names='value')



display(renderer)
display(HBox([checkbox_grid, checkbox_axes]))
util.rotate_animated(k, euler, "XYZ")
```

```python


rot_mat_x = np.array([[1, 0, 0],
    [0, np.cos(np.radians(-180)), -np.sin(np.radians(-180))],
    [0, np.sin(np.radians(-180)), np.cos(np.radians(-180))]
    ])

rot_mat_y = np.array([
    [np.cos(np.radians(-90)), 0, np.sin(np.radians(-90))],
    [0, 1, 0],
    [-np.sin(np.radians(-90)), 0, np.cos(np.radians(-90))]
    ])

rot_mat_z = np.array([
    [np.cos(np.radians(60)), -np.sin(np.radians(60)), 0],
    [np.sin(np.radians(60)),  np.cos(np.radians(60)), 0],
    [0,             0,             1]
    ])

rot_mat_ges = rot_mat_y @ rot_mat_x @ rot_mat_z
angles=[60, -180, -90]
order = "ZXY"

util.set_rotation(b, [0,0,0], order)
util.rotate_global_animated(b, angles, order)

util.set_rotation(b, [0,0,0], order)
util.rotate_animated(b, angles[::-1], order[::-1])

util.set_rotation(b, [0,0,0], order)
util.apply_rot_matrix_animated(b, rot_mat_ges)

print(rot_mat_ges, "Gesamtrotationsmatrix\n")

print(mat, "Rotation als Matrix vor der Drehung\n")

print(util.rot_matrix_to_euler(mat), "Rotation als ZYZ-Eulerwinkel\n")

print(rot_mat_ges @ mat, "Rotation als Matrix nach der Drehung\n")

print(util.rot_matrix_to_euler(rot_mat_ges @ mat, "ZYX"), "Roll Nick Gier Winkel")












```

```python
import pythreejs as three
from pythreejs import *
from ipywidgets import *
from IPython.display import display
import util


grid_group = util.create_grid_XY(32,0.5)
axes_group = util.create_axes(8)
env = util.Environment(frame=axes_group, grid=grid_group)

robot_group = three.Group()

wheel_height = 0.2
wheel_radius = 0.4
robot_radius = 1
chassis = util.create_cylinder([0, 0, wheel_radius], robot_radius, robot_radius, 0.5, 32, [0,255,0], True)

w0 = util.create_cylinder([0, 0, 1+wheel_height/2], wheel_radius, wheel_radius, wheel_height, 32, [255,0,0], True)
util.rotate(w0, [90,0,0], "XYZ")
w1 = util.create_cylinder([0, 0, -1-wheel_height/2], wheel_radius, wheel_radius, wheel_height, 32, [255,0,0], True)
util.rotate(w1, [90,0,0], "XYZ")

w0_axis = util.create_axes(2)
w1_axis = util.create_axes(2)

#w0.add([w0, w0_axis])
#w1.add([w1, w1_axis])

chassis.add([w0, w1])
util.rotate(chassis, [-90,0,0], "XYZ")
robot_axis = util.create_axes(2)
robot_group.add([chassis, robot_axis])

dummy = three.Group()
l1 = util.line_wheel_driven_robot(dummy, [0.01, 0, 0.00202], 1555)
l2 = util.line_wheel_driven_robot(dummy, [0.01, 0, 0.0], 990)

env.add([robot_group, l1, l2])

display(env)


```

```python
util.set_rotation(robot_group, [0,0,0], "XYZ")
util.set_translation(robot_group, [0,0,0])


util.move(robot_group, [0.01, 0.0, 0.00202], 1555)
util.move(robot_group, [0.01, 0.0, 0.0], 990)
    
    

```

```python
import pythreejs as three
from pythreejs import *
from ipywidgets import *
from IPython.display import display, Javascript
import numpy as np
import util
import time



def toggle_grid(change):
    grid_group.visible = not grid_group.visible

def toggle_axes(change):
    axes_group.visible = not axes_group.visible





width = 700
height = 500

scene = Scene()
scene.background = "#DDDDDD"


# Licht und Kamera
light = PointLight(color='white', intensity=1.5, position=[5, 5, 5])
camera = PerspectiveCamera(position=[8, 8, 8],aspect=width/height, fov=50)
camera.up = [0, 0, 1]

grid_group = util.create_grid_XY(32,0.5)
axes_group = util.create_axes(8)

robot_group = three.Group()

wheel_height = 0.2
wheel_radius = 0.4
robot_radius = 1
chassis = util.create_cylinder([0, 0, wheel_radius], robot_radius, robot_radius, 0.5, 32, [0,255,0], True)

w0 = util.create_cylinder([0, 0, 1+wheel_height/2], wheel_radius, wheel_radius, wheel_height, 32, [255,0,0], True)
util.rotate(w0, [90,0,0], "XYZ")
w1 = util.create_cylinder([0, 0, -1-wheel_height/2], wheel_radius, wheel_radius, wheel_height, 32, [255,0,0], True)
util.rotate(w1, [90,0,0], "XYZ")

w0_axis = util.create_axes(2)
w1_axis = util.create_axes(2)

#w0.add([w0, w0_axis])
#w1.add([w1, w1_axis])

chassis.add([w0, w1])
util.rotate(chassis, [-90,0,0], "XYZ")
robot_axis = util.create_axes(2)
robot_group.add([chassis, robot_axis])

dummy = three.Group()
l1 = util.line_wheel_driven_robot(dummy, [0.01, 0, 0.00202], 1555)
l2 = util.line_wheel_driven_robot(dummy, [0.01, 0, 0.0], 990)

scene.add([grid_group, axes_group, robot_group, light, l1, l2])
# Renderer mit Orbit-Steuerung
renderer = Renderer(camera=camera, scene=scene, controls=[OrbitControls(controlling=camera)], width=width, height=height, background_color="#87CEEB", background_opacity=1.0, antialias=True, precision='highp')


checkbox_grid = Checkbox(value=True, description='Show Grid')
checkbox_axes = Checkbox(value=True, description='Show Axes')


#interactive_control_scale = widgets.interactive(update_cube_scale, x=x_scale_slider, y=y_scale_slider, z=z_scale_slider)
checkbox_grid.observe(toggle_grid, names='value')
checkbox_axes.observe(toggle_axes, names='value')





display(renderer)
display(HBox([checkbox_grid, checkbox_axes]))


```

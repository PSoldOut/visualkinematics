```python
from pythreejs import *
import ipywidgets as widgets
from IPython.display import display

# Szene & Kamera
camera = PerspectiveCamera(position=[3, 3, 3], up=[0, 0, 1])
scene = Scene()

# Rechteck (Box)
box = BoxBufferGeometry(1, 1, 0.1)
material = MeshStandardMaterial(color='orange')
mesh = Mesh(geometry=box, material=material)

# Licht & Orbit Controls
light = DirectionalLight(position=[5, 5, 5])
scene.add(mesh, light)
controls = OrbitControls(controlling=camera)

# Add TransformControls from three.js via custom JavaScript
transform_controls = CustomThreeObject(
    """
    function(mesh, camera, renderer){
        let controls = new THREE.TransformControls(camera, renderer.domElement);
        controls.attach(mesh);
        controls.setMode("rotate");
        controls.size = 1.2;
        controls.showX = true;
        controls.showY = true;
        controls.showZ = true;
        return controls;
    }
    """,
    args=[mesh, camera]
)

scene.add(transform_controls)

# Renderer
renderer = Renderer(
    camera=camera,
    scene=scene,
    controls=[controls],
    background_color="#aaaaaa",
    antialias=True
)

display(renderer)
```
